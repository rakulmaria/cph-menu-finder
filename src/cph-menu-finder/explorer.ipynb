# Data Exploration

```python
from config import *
import pandas as pd
import numpy as np
```

### Load data

```python
path = WOLT_DATA_DIR / 'copenhagen'  
venues = pd.read_csv(path / 'venues.csv')
items = pd.read_csv(path / 'items.csv')
item_ingredients = pd.read_csv(path / 'item_ingredients.csv')
ingredients_raw = pd.read_csv(path / 'ingredients.csv').dropna()
ingredients_food = ingredients_raw[ingredients_raw['is_ingredient'] == 1.0]
```

### Merge it all together

First create the items to ingredients table

```python
item_ingredients = (
    item_ingredients
    .merge(ingredients_food[["ingredient_raw", "ingredient_verified"]], on="ingredient_raw")
    .rename(columns={"ingredient_verified": "ingredient"})[["item_id", "ingredient"]]
    .drop_duplicates() # "mayo" and "mayonnaise" in the same dish both become "mayonnaise"
)
```

Then merge with items and venues

```python
df = (
    item_ingredients
    .merge(items, on="item_id", validate="many_to_one")
    .merge(venues, on="wolt_id", validate="many_to_one")
)
df.drop(columns=['name']) # venue / name appears twice
```

### Explorative analysis

```python
dishes_per_ingredient = df.groupby("ingredient")["item_id"].nunique().sort_values(ascending=False)
dishes_per_ingredient.head()
```

```python
ingredient_in_venues = df.groupby("ingredient")["wolt_id"].nunique().sort_values(ascending=False)
ingredient_in_venues.head()
```

```python
venues_per_cuisine = venues.groupby("cuisine")["wolt_id"].nunique().sort_values(ascending=False)
venues_per_cuisine.head()
```

```python
counts = df.groupby(["cuisine", "ingredient"])["item_id"].nunique().unstack(fill_value=0)
share = pd.crosstab(df["cuisine"], df["ingredient"], normalize="index")
```
