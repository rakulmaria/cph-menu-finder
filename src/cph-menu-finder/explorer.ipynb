```python
from config import *
import pandas as pd
import numpy as np
```

```python
path = WOLT_DATA_DIR / 'copenhagen'  
df_ingredients_raw = pd.read_csv(path / 'ingredients.csv')
df_item_ingredients_raw = pd.read_csv(path / 'item_ingredients.csv')
df_items_raw = pd.read_csv(path / 'items.csv')
df_venues_raw = pd.read_csv(path / 'venues.csv')
```

```python
df_ingredients_raw.info()
```

```python
df_ingredients = df_ingredients_raw.dropna()
df_ingredients_food_only = df_ingredients[df_ingredients['is_ingredient'] == 1.0]

df_ingredients_food_only.head()
df_ingredients_food_only.info()
```

```python
df_item_ingredients_raw.info()
```

```python
item_to_ing = (
    df_item_ingredients_raw[['item_id', 'ingredient_raw']]
    .merge(df_ingredients_food_only[['ingredient_raw', 'ingredient_verified']], on='ingredient_raw')
    .drop(columns=['ingredient_raw'])
    .drop_duplicates()
)
```

```python
ing_to_dish = (
    df_items_raw
    .merge(item_to_ing, on='item_id')
)
ing_to_dish
```
