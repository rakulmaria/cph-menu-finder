```python
import spacy
spacy.prefer_gpu()
nlp = spacy.load("da_core_news_trf")
```

```python

doc = nlp("Sesambolle med fyld af oksebøf, dressing, ketchup, icebergsalat, løgringe, tomater og agurk.")

# lower casing
lower_token = [ tokens.text.lower() for tokens in doc ]

# stop words = unwanted words that does not give meaning to the sentence
stop_words = [ tokens.text for tokens in doc if not tokens.is_stop ]

# punctuation removal (! , . ?)
punctuation = [ tokens.text for tokens in doc if not tokens.is_punct ]

# lemmatization = convert text to its base or root form by cutting down tenses
lemma = [ tokens.lemma_ for tokens in doc ]
```
